```python
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_colwidth", 150)

print("Environnement JobPulse AI prêt.")
```

```python
#Définir le chemin des données
DATA_DIR = Path("../data/raw")

if not DATA_DIR.exists():
    raise FileNotFoundError(f"Dossier introuvable : {DATA_DIR.resolve()}")

files = list(DATA_DIR.glob("*.csv"))

print(f"Nombre de fichiers CSV trouvés : {len(files)}")

for file in files:
    size_mb = file.stat().st_size / (1024 * 1024)
    print(f"- {file.name} : {size_mb:.2f} MB")
```

```python
#Définir le chemin des données
jobs_2025 = pd.read_csv(
    DATA_DIR / "data_science_job_posts_2025.csv",
    low_memory=False
)

salaries = pd.read_csv(
    DATA_DIR / "ds_salaries.csv",
    low_memory=False
)

print("Dataset jobs_2025 :", jobs_2025.shape)
print("Dataset salaries :", salaries.shape)
```

```python
#Afficher les premières lignes
display(jobs_2025.head())
display(salaries.head())
```

```python
#Afficher les colonnes
print("Colonnes jobs_2025 :")
print(jobs_2025.columns.tolist())

print("\nColonnes salaries :")
print(salaries.columns.tolist())
```

```python
#Informations générales
print("Informations jobs_2025")
jobs_2025.info()

print("\nInformations salaries")
salaries.info()
```

```python
#Valeurs manquantes
missing_jobs = (
    jobs_2025.isnull()
    .sum()
    .sort_values(ascending=False)
)

missing_salaries = (
    salaries.isnull()
    .sum()
    .sort_values(ascending=False)
)

print("Valeurs manquantes dans jobs_2025 :")
display(missing_jobs[missing_jobs > 0])

print("Valeurs manquantes dans salaries :")
display(missing_salaries[missing_salaries > 0])
```

```python
#Doublons
print("Doublons jobs_2025 :", jobs_2025.duplicated().sum())
print("Doublons salaries :", salaries.duplicated().sum())
```

```python
#Échantillon de postings.csv
postings_sample = pd.read_csv(
    DATA_DIR / "postings.csv",
    nrows=1000,
    low_memory=False
)

print("Dimensions de l'échantillon :", postings_sample.shape)
print("Colonnes :", postings_sample.columns.tolist())

display(postings_sample.head())
```

```python
#Compter les lignes de postings.csv par morceaux
total_rows = 0

for chunk in pd.read_csv(
    DATA_DIR / "postings.csv",
    chunksize=100_000,
    low_memory=False
):
    total_rows += len(chunk)

print("Nombre total de lignes :", total_rows)
```

```python
#Créer un échantillon de travail
postings_work = pd.read_csv(
    DATA_DIR / "postings.csv",
    nrows=50_000,
    low_memory=False
)

print("Échantillon de travail :", postings_work.shape)
```

```python
#Enregistrer cet échantillon
PROCESSED_DIR = Path("../data/processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

postings_work.to_csv(
    PROCESSED_DIR / "postings_sample_50000.csv",
    index=False
)

print("Échantillon enregistré.")
```

```python
#Résumé automatique des quatre datasets
datasets_summary = []

datasets = {
    "jobs_2025": jobs_2025,
    "salaries": salaries,
    "resume_sample": resume_sample,
    "postings_sample": postings_sample
}

for name, df in datasets.items():
    datasets_summary.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicates": int(df.duplicated().sum()),
        "missing_values": int(df.isnull().sum().sum())
    })

summary_df = pd.DataFrame(datasets_summary)

display(summary_df)
```
